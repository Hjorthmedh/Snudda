# Change identify of centre neuron

This code creates a network, and changes the neuron_path, parameter_key and morphology key of the neuron.

```python
import os
import snudda

bg_path = "../../../BasalGangliaData/data/"
if os.path.isdir(bg_path):
    snudda_data = bg_path
else:
    snudda_data = os.path.join(os.path.dirname(snudda.__file__), "data")

number_of_neurons = 50
```

```python
import os

network_path = os.path.join("networks/change_neuron")

from snudda import SnuddaInit
struct_def = {"Striatum": number_of_neurons }

si = SnuddaInit(network_path=network_path, struct_def=struct_def, random_seed=123, snudda_data=snudda_data)
```

```python
from snudda import Snudda
snudda = Snudda(network_path=network_path)
snudda.place_neurons()
```

```python
from snudda.utils import SnuddaLoad
pos_file = os.path.join(network_path,"network-neuron-positions.hdf5")
sl_old = SnuddaLoad(pos_file)
```

```python
centre_neuron_idx, dist_to_centre = list(sl_old.get_centre_neurons_iterator(n_neurons=1, neuron_type="dSPN"))[0]
```

```python
centre_neuron_idx, dist_to_centre
```

```python
sl_old.data["neurons"][centre_neuron_idx]
```

```python
from snudda.utils.reposition_neurons import RepositionNeurons
rn = RepositionNeurons(pos_file)
```

```python
rn.set_neuron_path(centre_neuron_idx,
                   os.path.join("../../../BasalGangliaData/data/", 
                                "neurons/striatum/dspn/str-dspn-e150602_c1_D1-mWT-0728MSN01-v20220620/"))
rn.set_parameter_key(centre_neuron_idx, "p09df1da3")
rn.set_morphology_key(centre_neuron_idx, "m2bd5d73d")
rn.update_morphology(centre_neuron_idx)
rn.close()
```

```python
sl_new = SnuddaLoad(pos_file)
```
